```python
Sales = int(input("Enter sales ($): "))
TaxRate = int(input("Enter tax rate (%): "))
Cogs = Sales*0.4
Profit = Sales - Cogs
Taxes = Profit * (TaxRate/100)
NetIncome = Profit - Taxes
print(f"Sales: ${Sales}\nCost of Goods Sold: ${Cogs}\nGross Profit: ${Profit}\nTaxes: {Taxes}\nNet Income: {NetIncome}")
```

```python
print("This is a numerical calculator. You will be prompted for 2 numbers and a mathematical function (+ - * / // % **).")
num1 = float(input("What is the first number: "))
num2 = float(input("What is the second number: "))
operator = input("What is the operation (+ - * / // % **): ")

if "+" == operator:
  print(num1 + num2)
elif "-" == operator:
  print(num1 - num2)
elif  "**" == operator:
  print(num1**num2)
elif "//" == operator:
  print(num1//num2)
elif "%" == operator:
  print(num1%num2)
elif "*" == operator:
  print(num1*num2)
elif "/" == operator:
  print(round(num1/num2,2))
else:
  print("Whatever was entered is not a valid operator.")
```
