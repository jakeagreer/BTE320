```python
# cogs calculator

Sales = int(input("Enter sales($): "))
Cogs = Sales*0.4
Profit = Sales - Cogs
print(f"Sales: ${Sales}\nCost of Goods Sold: ${Cogs}\nGross Profit: ${Profit}")
```

```python
# tax rate calculator

Sales = int(input("Enter sales ($): "))
TaxRate = int(input("Enter tax rate (%): "))
Cogs = Sales*0.4
Profit = Sales - Cogs
Taxes = Profit * (TaxRate/100)
NetIncome = Profit - Taxes
print(f"Sales: ${Sales}\nCost of Goods Sold: ${Cogs}\nGross Profit: ${Profit}\nTaxes: {Taxes}\nNet Income: {NetIncome}")
```

```python
# if statement calculator

print("This is a numerical calculator. You will be prompted for 2 numbers and a mathematical function (+ - * / // % **).")
num1 = float(input("What is the first number: "))
num2 = float(input("What is the second number: "))
operator = input("What is the operation (+ - * / // % **): ")

if "+" == operator:
  print(num1 + num2)
elif "-" == operator:
  print(num1 - num2)
elif  "**" == operator:
  print(num1**num2)
elif "//" == operator:
  print(num1//num2)
elif "%" == operator:
  print(num1%num2)
elif "*" == operator:
  print(num1*num2)
elif "/" == operator:
  print(round(num1/num2,2))
else:
  print("Whatever was entered is not a valid operator.")
```

```python
# for loop hourly parking rates

base = 5
hr = 2.50
for i in range(1,9,1):
  payment = base + (i*hr)
  if payment < 10:
    print(f"{i} 10.0")
  elif payment > 20:
    print(f"{i} 20.0")
  else:
    print(i, payment)
```

```python
# movie theater profit
import math

fixed = 200
print("Ad$ Profit")
for ad in range(0,201,25):
  additional = 2 * round(math.sqrt(ad))
  earnings = (20 + additional) * 10
  profit = earnings - fixed - ad
  print(ad, profit)
```

```python
# while loop hourly parking rates

i = 1
base = 5
hr = 2.50
while i < 9:
  payment = base + (i * hr)
  if payment < 10:
    print(f"{i} 10.0")
  elif payment > 20:
    print(f"{i} 20.0")
  else:
    print(i, payment)
  i += 1
```

```python
# while loop movie theater profit
import math

fixed = 200
i = 0
print("Ad$ Profit")
while i <= 200:
  additional = 2 * round(math.sqrt(i))
  earnings = (20 + additional) * 10
  profit = earnings - fixed - i
  print(i, profit)
  i += 25
```

```python
#Rock Paper Scissors
import random

i = input("Do you want to play Rock Paper Scissors (y/n): ")
player = 0
computer = 0
while i == "y":
  choice = input(f"Currently the score of your game is Player: {player} vs. Computer: {computer} \nDo you pick Rock, Paper, or Scissors: ")
  computer_choice = random.choice(["Rock", "Paper", "Scissors"])
  if choice == computer_choice:
    print(f"You both picked {choice}. It is a tie.")
  elif ((choice == "Rock") & (computer_choice == "Paper")) | ((choice == "Paper") & (computer_choice == "Scissors")) | ((choice == "Scissors") & (computer_choice == "Rock")):
    print(f"You lose. The computer had {computer_choice}.")
    computer += 1
  else:
    print(f"You won. The computer had {computer_choice}")
    player += 1
  i = input(f"It is now Player: {player} vs. Computer: {computer}\nDo you want to play again (y/n): ")
print("Thanks for playing!")
```
