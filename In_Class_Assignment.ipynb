```python
# if statement calculator

print("This is a numerical calculator. You will be prompted for 2 numbers and a mathematical function (+ - * / // % **).")
num1 = float(input("What is the first number: "))
num2 = float(input("What is the second number: "))
operator = input("What is the operation (+ - * / // % **): ")

if "+" == operator:
  print(num1 + num2)
elif "-" == operator:
  print(num1 - num2)
elif  "**" == operator:
  print(num1**num2)
elif "//" == operator:
  print(num1//num2)
elif "%" == operator:
  print(num1%num2)
elif "*" == operator:
  print(num1*num2)
elif "/" == operator:
  print(round(num1/num2,2))
else:
  print("Whatever was entered is not a valid operator.")
```

```python
# movie theater profit
import math

fixed = 200
print("Ad$ Profit")
for ad in range(0,201,25):
  additional = 2 * round(math.sqrt(ad))
  earnings = (20 + additional) * 10
  profit = earnings - fixed - ad
  print(ad, profit)
```

```python
# Rock Paper Scissors
import random

i = input("Do you want to play Rock Paper Scissors (y/n): ")
player = 0
computer = 0
while i == "y":
  choice = input(f"Currently the score of your game is Player: {player} vs. Computer: {computer} \nDo you pick Rock, Paper, or Scissors: ")
  computer_choice = random.choice(["Rock", "Paper", "Scissors"])
  if choice == computer_choice:
    print(f"You both picked {choice}. It is a tie.")
  elif ((choice == "Rock") & (computer_choice == "Paper")) | ((choice == "Paper") & (computer_choice == "Scissors")) | ((choice == "Scissors") & (computer_choice == "Rock")):
    print(f"You lose. The computer had {computer_choice}.")
    computer += 1
  else:
    print(f"You won. The computer had {computer_choice}")
    player += 1
  i = input(f"It is now Player: {player} vs. Computer: {computer}\nDo you want to play again (y/n): ")
print("Thanks for playing!")
```

```python
def stringR(s):
  if len(s) < 2:
    return s
  else:
    return s[-1] + stringR(s[1:-1]) + s[0]

def main():
  s = input("Give me a word to reverse recursively: ")
  sr = stringR(s)
  print(f"{sr} is {s} reversed")


main()
```

```python
#Part A
def roi(cost, revenue):
  assert cost != 0
  assert isinstance(cost, (int, float))
  assert isinstance(revenue, (int, float))
  roi = (revenue - cost) / cost
  return roi

def main():
  cost = int(input("What is the cost: "))
  revenue = int(input("What is the revenue: "))
  print(roi(cost, revenue))


main()
```

```python
#Part B
def loan_payment(principal, annual_rate, months):
  assert isinstance(months, (int, float))
  assert isinstance(principal, (int, float))
  assert isinstance(annual_rate, (int, float))

  if principal <= 0 or months <= 0 or annual_rate < 0:
    raise ValueError("Principal, Months, and Annual Rate must be positive numbers")

  try:
    r = annual_rate/12
    if r == 0:
      pmt = principal/months
    else:
      pmt = r * principal / (1-(1+r)**(-months))
    return pmt
  except Exception as e:
    raise Exception(f"Error: {e}")

def main():
  p = float(input("What is the principal ammount: "))
  a = float(input("What is the annual rate (in decimal form): "))
  m = int(input("How many months: "))
  print(f"Your Loan Payment is: ${round(loan_payment(p,a,m), 2)}")


main()
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(self.first)

  def __str__(self):
    return f"{self.first} {self.last}"


def main():
  name = Greeting("Jake", "Greer")
  name.display(True)
  name.display()
  print(name)

main()
```
