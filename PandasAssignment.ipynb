````python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams.update({'font.size': 20, 'figure.figsize': (10, 8)})
df = pd.read_csv('travel-times.csv')
````

Print the shape of the dataset

````python
print("Shape of the dataset:", df.shape)
````

Print the first 15 rows of the dataset

````python
print(df.head(15))
df.head(15)
````

Get information for the features (columns) with missing values

````python
df.info()
````

Drop duplicate values (if any) by keeping only the last instance

````python
df = df.drop_duplicates(keep='last')
df
````

Calculate the total number of missing values (if any) on each column

````python
print(df.isnull().sum())
````

Create two copies of the dataframe and drop missing values

````python
df_copy1 = df.copy()
df_copy2 = df.copy()

df_copy1 = df_copy1.dropna(axis=0)

df_copy2 = df_copy2.dropna(axis=1)
````

Get summary statistics and see the correlation between the numerical columns

````python
display(df.describe())

display(df.corr(numeric_only=True))
````

Show rows 11 to 14

````python
df.iloc[11:15]
````

Create a subset with trips occurred on November 23, 2011 and January 6, 2012

````python
dates_to_find = ['11/23/2011', '1/6/2012', '2011-11-23', '2012-01-06']
subset_df = df[df['Date'].isin(dates_to_find)]

display(subset_df)
````

Produce a scatterplot between Distance and TotalTime

````python
df.plot(kind='scatter', x='Distance', y='TotalTime', title='Distance vs. TotalTime')
plt.show()
````

Produce boxplots for AvgSpeed and AvgMovingSpeed

````python
df.boxplot(column='AvgSpeed')
plt.title('Boxplot of Average Speed')
plt.ylabel('Speed')
plt.show()
````

````python
df.boxplot(column='AvgMovingSpeed')
plt.title('Boxplot of Average Moving Speed')
plt.ylabel('Speed')
plt.show()
````
